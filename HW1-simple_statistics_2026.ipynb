## Simple statistics

Okay, now for some simple statistics to make sure you're computing environment is setup and to practice a few commands we'll use over and over again.

First, let's draw a 100k points from a Gaussian distribution. In many areas of statistics a Gaussian distribution is called the 'Normal distribution'. So we'll dig into the `stats` package from `scipy` (see `import` lines above). Read up on the command to make sure you understand what it is doing.

```python
d = stats.norm.rvs(loc = 5.0, scale = .01, size = 1000000)
```

Now let's plot a histogram of these points. *Make sure you understand what each of these commands is doing.*

```python
fig, ax = plt.subplots(1, 1)
ax.hist(d,50, density=True)
plt.tick_params(labelsize = 24)
plt.xlim([4.95,5.05])
x = np.linspace(4.95,5.05,1000)
ax.plot(x,stats.norm.pdf(x,loc = 5., scale = 0.01),linewidth = 8,alpha = 0.7)
plt.show()
```

```python
ax = plt.hist(d,50, density=True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([4.95,5.05])
plt.show()
```

```python
fig, ax = plt.subplots(1, 1)
ax.hist(d,50, density=True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([4.95,5.05])
x = np.linspace(4.95,5.05,1000)
ax.plot(x,stats.norm.pdf(x,loc = 5., scale = 0.01),linewidth = 8,alpha = 0.7)
plt.show()
```

We will now draw points from and visualize a Gaussian distribution centered at 0 with standard deviation 0.1. This will, in large part, mirror the tutorial above and both the regular and log-scale visualizations will have the fitted curve overlaid on top.

```python
dis = stats.norm.rvs(loc = 0.0, scale = .1, size = 1000000)
```

Now to plot a regular histogram of this...

```python
fig, ax = plt.subplots(1, 1)
ax.hist(dis,50, density=True)
plt.tick_params(labelsize = 24)
plt.xlim([-0.5,0.5])
x = np.linspace(-0.5,0.5,1000)
ax.plot(x,stats.norm.pdf(x,loc = 0., scale = 0.1),linewidth = 8,alpha = 0.7)
plt.show()
```

In log scale:

```python
fig, ax = plt.subplots(1, 1)
ax.hist(dis,50, density=True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([-0.5,0.5])
x = np.linspace(-0.5,0.5,1000)
ax.plot(x,stats.norm.pdf(x,loc = 0., scale = 0.1),linewidth = 8,alpha = 0.7)
plt.show()
```
